## Code

```python
import json
from pathlib import Path

def build():
    S = {
        "diab_wo_ckd": set(diab_wo_ckd),
        "ckd1": set(diab_ckd1),
        "ckd2": set(diab_ckd2),
        "ckd3": set(diab_ckd3),
        "ckd4": set(diab_ckd4),
        "nondiab_no_ckd": set(non_diabetes_patients),
        "all_ckd": set(non_diabetes_patients_ckd),
    }

    early = S["ckd1"] | S["ckd2"]          # stage 1-2, all diabetic
    severe = S["ckd3"] | S["ckd4"]         # stage 3-4, all diabetic
    no_ckd = S["nondiab_no_ckd"] | S["diab_wo_ckd"]
    all_ckd = S["all_ckd"] | early         # includes stage-unrecorded positives
    diabetic = S["diab_wo_ckd"] | early | severe

    tasks = {
        "main_any_ckd": {
            "pos": all_ckd, "neg": no_ckd,
            "note": "PRIMARY: CKD at any stage vs no CKD",
        },
        "check_diabetic": {
            "pos": (early | severe) & diabetic,
            "neg": S["diab_wo_ckd"],
            "note": "confound check: diabetic patients only",
        },
        "check_nondiabetic": {
            "pos": all_ckd - diabetic,
            "neg": S["nondiab_no_ckd"],
            "note": "confound check: non-diabetic patients only",
        },
        "legacy_severe_only": {
            "pos": S["all_ckd"], "neg": no_ckd,
            "note": "reference: stage 3-4 vs no CKD, as in the draft paper",
        },
    }

    files = set()
    root = Path(DATA_ROOT)
    if root.is_dir():
        files = {p.stem.replace(FILE_PREFIX, "") for p in root.glob("*.tsv")}
        print(f"TSV files on disk: {len(files)}\n")
    else:
        print(f"(DATA_ROOT not found: {root} --- skipping coverage report)\n")

    out = {}
    for name, spec in tasks.items():
        pos, neg = spec["pos"], spec["neg"]
        overlap = pos & neg
        if overlap:
            print(f"!! {name}: {len(overlap)} patients in BOTH classes; "
                  f"removing from positives")
            pos = pos - neg

        labels = {p: 1 for p in sorted(pos)}
        labels.update({p: 0 for p in sorted(neg)})

        n_pos, n_neg = len(pos), len(neg)
        diab_pos = len(pos & diabetic)
        diab_neg = len(neg & diabetic)

        line = (f"{name:20s} n={n_pos + n_neg:4d}  "
                f"pos={n_pos:4d} neg={n_neg:4d}  "
                f"pos_rate={n_pos / (n_pos + n_neg):.1%}")
        if files:
            with_data = {p for p in labels if p in files}
            wd_pos = sum(labels[p] for p in with_data)
            line += (f"   with data: {len(with_data):4d} "
                     f"({wd_pos} pos / {len(with_data) - wd_pos} neg)")
        print(line)
        print(f"{'':20s} diabetic share: pos "
              f"{diab_pos / max(n_pos, 1):.0%}, neg {diab_neg / max(n_neg, 1):.0%}"
              + ("   <-- confounded" if abs(diab_pos / max(n_pos, 1)
                                            - diab_neg / max(n_neg, 1)) > 0.15
                 else ""))
        print(f"{'':20s} {spec['note']}\n")

        dest = Path(OUT_DIR) / f"labels_{name}.json"
        with open(dest, "w") as fh:
            json.dump({"task": name, "note": spec["note"],
                       "n_positive": n_pos, "n_negative": n_neg,
                       "labels": labels}, fh, indent=1)
        out[name] = str(dest)

    print("Written:")
    for name, path in out.items():
        print(f"  {path}")
```

```python
import json
import re
from pathlib import Path

def parse(b):
    return set(re.findall(r"P\d{4}", b))

def build():
    Path(OUT_DIR).mkdir(parents=True, exist_ok=True)

    d_no, n_no = parse(TAB_DIABETES), parse(TAB_NONDIABETES)
    ctrl = d_no | n_no
    d_e = parse(TAB_DIABETES_ECKD) - ctrl
    d_m = parse(TAB_DIABETES_MCKD) - ctrl
    n_e = parse(TAB_NONDIABETES_ECKD) - ctrl
    n_m = parse(TAB_NONDIABETES_MCKD) - ctrl - d_m

    if not d_no or not n_e:
        print("Paste the spreadsheet tabs into the LISTS section first.")
        return

    diabetic = d_no | d_e | d_m
    ckd = d_e | d_m | n_e | n_m
    no_ckd = ctrl
    mckd, eckd = d_m | n_m, d_e | n_e

    print("COHORT AS PARSED")
    print(f"{'':14s} {'no CKD':>8} {'eCKD':>8} {'mCKD':>8}")
    print(f"{'Diabetes':14s} {len(d_no):8d} {len(d_e):8d} {len(d_m):8d}")
    print(f"{'nonDiabetes':14s} {len(n_no):8d} {len(n_e):8d} {len(n_m):8d}")

    root = Path(DATA_ROOT)
    files = ({p.stem.replace(FILE_PREFIX, "") for p in root.glob("*.tsv")}
             if root.is_dir() else set())
    print(f"\nTSV files on disk: {len(files) if files else 'DATA_ROOT not found'}\n")

    tasks = {
        "severity_nondiabetic": {
            "pos": n_m, "neg": n_e,
            "note": "PRIMARY. Moderate-to-severe versus early CKD, non-diabetic "
                    "patients only. Patient ID carries no information "
                    "(AUC 0.519, p = 0.64) and diabetes is absent from both "
                    "groups. Largest confound-free cohort in the dataset.",
        },
        "severity_all": {
            "pos": mckd, "neg": eckd,
            "note": "Severity across the whole cohort. Clean on enrollment "
                    "order (ID AUC 0.535) but diabetes is unevenly distributed "
                    "(45% of mCKD versus 25% of eCKD), so a positive result "
                    "could reflect diabetes. Report beside the non-diabetic "
                    "version, not instead of it.",
        },
        "severity_diabetic": {
            "pos": d_m, "neg": d_e,
            "note": "Severity within diabetic patients. Confounded with "
                    "enrollment order (ID AUC 0.632); interpret with care.",
        },
        "dkd_vs_nondiabetic_mckd": {
            "pos": d_m, "neg": n_m,
            "note": "Diabetic versus non-diabetic moderate-to-severe CKD. Clean "
                    "on enrollment order (ID AUC 0.565). Both groups have "
                    "kidney disease, so this detects diabetes, not CKD.",
        },
        "detect_any_ckd": {
            "pos": ckd, "neg": no_ckd,
            "note": "CONFOUNDED (ID AUC 0.719). Retained for comparison with "
                    "the draft paper and for the ID-matched sensitivity "
                    "analysis, not as a primary result.",
        },
        "detect_mckd": {
            "pos": mckd, "neg": no_ckd,
            "note": "CONFOUNDED (ID AUC 0.765). Closest to the definition used "
                    "in the draft paper.",
        },
        "detect_nondiabetic": {
            "pos": n_e | n_m, "neg": n_no,
            "note": "CONFOUNDED (ID AUC 0.695). Diabetes removed, enrollment "
                    "confound remains.",
        },
        "positive_control_diabetes": {
            "pos": diabetic, "neg": (ckd | no_ckd) - diabetic,
            "note": "Diabetes versus no diabetes. Establishes that the pipeline "
                    "detects real biology. In this expanded cohort the contrast "
                    "is itself enrollment-confounded (ID AUC 0.643); the clean "
                    "version used the 434-patient subset.",
        },
    }

    print(f"{'task':28s} {'n':>5} {'pos':>5} {'neg':>5} {'pos%':>6}"
          + ("   with data (pos/neg)" if files else ""))
    print("-" * (80 if files else 54))
    for name, spec in tasks.items():
        pos, neg = spec["pos"] - spec["neg"], spec["neg"] - spec["pos"]
        labels = {p: 1 for p in sorted(pos)}
        labels.update({p: 0 for p in sorted(neg)})
        n = len(labels)
        line = (f"{name:28s} {n:5d} {len(pos):5d} {len(neg):5d} "
                f"{len(pos)/max(n,1):6.1%}")
        if files:
            wd = {p for p in labels if p in files}
            wp = sum(labels[p] for p in wd)
            line += f"   {len(wd):4d} ({wp}/{len(wd)-wp})"
        print(line)
        with open(Path(OUT_DIR) / f"labels_{name}.json", "w") as fh:
            json.dump({"task": name, "note": spec["note"],
                       "n_positive": len(pos), "n_negative": len(neg),
                       "labels": labels}, fh, indent=1)

    print(f"\n{len(tasks)} label files written to {OUT_DIR}/")
```
